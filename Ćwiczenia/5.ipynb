
#### Zadanie 1 – Tworzenie i indeksowanie listy



Stwórz listę zawierającą pomiary temperatury z 7 dni: `[22.5, 23.0, 21.5, 24.0, 23.5, 22.0, 23.0]`. Wyświetl:

- Temperaturę z pierwszego dnia

- Temperaturę z ostatniego dnia

- Temperatury z trzech środkowych dni (slicing)

```python
# Zadanie 1
temp = [22.5, 23.0, 21.5, 24.0, 23.5, 22.0, 23.0]

print(f"""Temperatura pierwszego dnia: {temp[0]}
Temperatura ostatniego dnia: {temp[-1]}
Temperatury trzech środkowych dni: {temp[2:5]}""")
```

#### Zadanie 3 – Sortowanie listy
Masz listę wyników: `[0.78, 0.92, 0.85, 0.88, 0.81]`. Posortuj ją malejąco i wyświetl 3 najlepsze wyniki.

```python
# Zadanie 3
results = [0.78, 0.92, 0.85, 0.88, 0.81]
sorted_results = sorted(results, reverse=True)

print(f"3 najlepsze wyniki: {sorted_results[:3]}")
```

```python
# Zadanie 10
dataset_a = {'wiek', 'plec', 'miasto', 'dochod', 'wyksztalcenie', 'stan_cywilny'}
dataset_b = {'wiek', 'plec', 'miasto', 'pensja', 'stanowisko', 'staz'}

cechy_wspólne = dataset_a & dataset_b
print(f"Cechy wspólne: {cechy_wspólne}")

tylko_a = dataset_a - dataset_b
print(f"Cechy tylko w dataset_a: {tylko_a}")

tylko_b = dataset_b - dataset_a
print(f"Cechy tylko w dataset_b: {tylko_b}")

wszystkie_cechy = dataset_a | dataset_b
print(f"Wszystkie unikalne cechy: {wszystkie_cechy}")

roznica_symetryczna = dataset_a ^ dataset_b
print(f"Różnica symetryczna: {roznica_symetryczna}")

test_subset = {'wiek', 'plec'}
czy_podzbiorem = test_subset.issubset(cechy_wspólne)
print(f"Czy {test_subset} jest podzbiorem cech wspólnych: {czy_podzbiorem}")
```

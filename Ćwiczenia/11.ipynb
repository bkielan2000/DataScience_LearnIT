| Niemcy    | 4      | 4        | 3     |



Dodaj kolumnę `Bilans` = Bramki − Stracone.

Wypisz kraj z najlepszym bilansem i posortuj tabelę malejąco po tej kolumnie.



---

```python
reprezentacje = pd.DataFrame({
    'Kraj': ['Brazylia', 'Francja', 'Polska', 'Argentyna', 'Niemcy'],
    'Bramki': [7, 5, 3, 6, 4],
    'Stracone': [2, 3, 4, 1, 4],
    'Mecze': [3, 3, 3, 3, 3]
})

reprezentacje['Bilans'] = reprezentacje['Bramki'] - reprezentacje['Stracone']

# Znalezienie kraju z najlepszym bilansem
najlepszy_bilans = reprezentacje.loc[reprezentacje['Bilans'].idxmax()]
print(f"Kraj z najlepszym bilansem: {najlepszy_bilans['Kraj']} z bilansem {najlepszy_bilans['Bilans']}")

# Sortowanie tabeli malejąco po kolumnie Bilans
reprezentacje_posortowana = reprezentacje.sort_values('Bilans', ascending=False)
print("\nTabela posortowana po bilansie (malejąco):")
print(reprezentacje_posortowana)
```

## Zadanie 2



Z Titanica wyciągnij kobiety z 1. lub 2. klasy **dwoma sposobami**:

- raz boolean indexingiem

- raz przez `.query()`



Sprawdź czy obie metody dają identyczny wynik porównując `.shape`.

Na koniec: jaki jest survival rate tej grupy?

```python
titanic_raw.head()
```

```python
kobiety_klasa_1_2_bool = titanic_raw[(titanic_raw['Sex'] == 'female')
    & ((titanic_raw['Pclass'] == 1) | (titanic_raw['Pclass'] == 2))]

kobiety_klasa_1_2_query = titanic_raw.query("Sex == 'female' and (Pclass == 1 or Pclass == 2)")

print(f"Boolean indexing shape: {kobiety_klasa_1_2_bool.shape}")
print(f"Query shape: {kobiety_klasa_1_2_query.shape}")

survival_rate = kobiety_klasa_1_2_bool['Survived'].mean()
print(f"Survival rate dla kobiet z 1. lub 2. klasy: {survival_rate:.2%}")
```

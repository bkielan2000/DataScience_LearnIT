### Zadanie 8 — Wypełnianie braków grupowo

W datasecie Titanic wypełnij brakujące wartości wieku medianą według klasy.

Wymagania:
- Sprawdź liczbę braków w `Age`
- Wypełnij medianą grupową (według `Pclass`) — **poprawny wzorzec Pandas 3.0**
- Sprawdź, czy braki zostały usunięte
- Porównanie przed/po

*(proste)*

````python
titanic = titanic_raw.copy()

titanic['Age'] = titanic['Age'].fillna(titanic.groupby('Pclass')['Age'].transform('median'))

print(f"Przed: {titanic_raw['Age'].isna().sum()} | Po: {titanic['Age'].isna().sum()}")
````

## Zadania średnie (9–12)

### Zadanie 9 — Multi-Level Aggregation

Oblicz średni wiek i survival rate według klasy, płci i czy pasażer podróżował sam.

Wymagania:
- Stwórz kolumnę `Is_Alone` (czy `SibSp + Parch == 0`)
- Grupuj według 3 kolumn: `Pclass`, `Sex`, `Is_Alone`
- Oblicz średnią wieku i survival rate
- Spłaszcz MultiIndex

*(średnie)*

````python
titanic_survival = titanic_raw.copy()
titanic_survival['Is_Alone'] = (titanic_survival['SibSp'] + titanic_survival['Parch'] == 0)

survival_rate = titanic_survival.groupby(['Pclass', 'Sex', 'Is_Alone']).agg({'Age': 'mean', 'Survived': 'mean'}).reset_index()
survival_rate
````

Zadanie agg & transform: Dodaj kolumnę Udzial_Proc pokazującą, jaki procent sprzedaży swojego regionu stanowi dany wiersz.

(proste)

````python
sales = pd.DataFrame({
    'Region': np.random.choice(['A', 'B', 'C'], 100),
    'Sprzedaż': np.random.randint(50, 500, 100)
})

sales['Udzial_Proc'] = sales.agg({'Sprzedaż': lambda x: x / x.sum() * 100})['Sprzedaż']
sales.head()
````

**Zadanie Feature Engineering:** Stwórz binarną cechę `Ma_Kabine` (czy pasażer miał przypisaną kabinę, czyli

`Cabin` nie jest brakiem danych) i sprawdź survival rate w podziale na nią.

````python
titanic_has_cabin = titanic_raw.copy()
titanic_has_cabin['Ma_Kabine'] = titanic_has_cabin['Cabin'].notnull()
titanic_has_cabin_surval_rate = titanic_has_cabin.groupby('Ma_Kabine')['Survived'].mean()
print(titanic_has_cabin_surval_rate)
````

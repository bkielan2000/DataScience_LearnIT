### Zadanie 6 — Boolean indexing

Masz tablicę z oceną studentów:
```python
grades = np.array([78, 45, 92, 67, 34, 88, 56, 91, 23, 85, 73, 95])
```

Wykonaj:
- Znajdź wszystkie oceny >= 80 (oceny wysokie)
- Znajdź wszystkie oceny < 50 (oceny niedostateczne)
- Policz ile jest ocen w przedziale [60, 80)
- Zmień wszystkie oceny < 50 na 50 (podniesienie do minimum)

*(proste)*

````python
grades = np.array([78, 45, 92, 67, 34, 88, 56, 91, 23, 85, 73, 95])

oceny_wysokie = grades[grades >= 80]
oceny_niedostateczne = grades[grades < 50]
oceny_w_przedziale = np.sum((grades >= 60) & (grades < 80))

zmienione_oceny = np.copy(grades)
zmienione_oceny = np.clip(zmienione_oceny, 50, None)

print(f"""
Oceny wysokie (>= 80): {oceny_wysokie}
Oceny niedostateczne (< 50): {oceny_niedostateczne}
Liczba ocen w przedziale [60, 80): {oceny_w_przedziale}
Oceny po podniesieniu do minimum 50: {zmienione_oceny}
""")
````

### Zadanie 10 — Broadcasting w praktyce

Dana tablica ocen (5 studentów x 4 egzaminy):
```python
exams = np.array([
    [78, 85, 92, 88],
    [65, 72, 68, 70],
    [90, 88, 95, 92],
    [55, 60, 58, 62],
    [82, 79, 85, 80]
])
```

Dodaj bonus do każdego egzaminu: `[5, 10, 5, 10]`

Następnie pomnóż wyniki każdego studenta przez mnożnik: `[1.0, 1.1, 0.95, 1.15, 1.0]`

**Wymagania:**
- Użyj broadcastingu (bez pętli)
- Wypisz wyniki przed i po

*(średnie)*

````python
exams = np.array([
    [78, 85, 92, 88],
    [65, 72, 68, 70],
    [90, 88, 95, 92],
    [55, 60, 58, 62],
    [82, 79, 85, 80]
])

bonus = np.array([5, 10, 5, 10])
exams_w_bonus = np.copy(exams)
exams_w_bonus = exams_w_bonus + bonus

mnoznik = np.array([1.0, 1.1, 0.95, 1.15, 1.0])
exams_mnozone = exams_w_bonus * mnoznik.reshape(5, 1)

print("Wyniki przed dodaniem bonusu:\n", exams)
print("\nWyniki po dodaniu bonusu:\n", exams_w_bonus)
print("\nWyniki po dodaniu bonusu i zastosowaniu mnożnika:\n", exams_mnozone)
````

# Ćwiczenie — trzy biblioteki wizualizacji

Jeden prosty wykres w każdej z trzech bibliotek: **Matplotlib**, **Seaborn**, **Plotly**.
Pracujemy na tym samym zbiorze danych (napiwki w restauracji), żeby łatwo porównać podejścia.

## Setup — uruchom raz

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

tips = sns.load_dataset('tips')
print(tips.shape)
tips.head()
```

---
### Zadanie 1 — Matplotlib

Narysuj **histogram** kolumny `total_bill` (rachunek).

**Wskazówki:**
- `fig, ax = plt.subplots()`
- `ax.hist(dane, bins=20)`
- dodaj `ax.set_title()` i `ax.set_xlabel()`

```python
# Napisz tutaj swój kod
total_bill = tips['total_bill']

fig, ax = plt.subplots()
ax.hist(total_bill, bins=20)
ax.set_title('Histogram total_bill')
ax.set_xlabel('Total Bill')
```

---
### Zadanie 2 — Seaborn

Narysuj **wykres pudełkowy (box plot)**: `total_bill` na osi Y, `day` (dzień tygodnia) na osi X.

**Wskazówki:**
- `sns.boxplot(data=tips, x='day', y='total_bill')`

```python
# Napisz tutaj swój kod
total_bill = tips['total_bill']

sns.boxplot(data=tips, x='day', y='total_bill')
```

---
### Zadanie 3 — Plotly

Narysuj **interaktywny scatter plot**: `total_bill` vs `tip`, kolorowany według `time` (pora dnia).

**Wskazówki:**
- `px.scatter(tips, x='total_bill', y='tip', color='time')`
- na końcu wywołaj `.show()`

```python
# Napisz tutaj swój kod
interactive_bill_tip = px.scatter(tips, x='total_bill', y='tip', color='time')
interactive_bill_tip.show()
```

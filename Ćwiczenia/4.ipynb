```python
# 1. Kalkulator bmi
waga = 70
wzrost = 1.75

bmi =  waga / (wzrost**2)
print(f"BMI: {bmi:.2f} kg/m^2")

if bmi < 18.5:
    print("Niedowaga")
elif bmi >= 25:
    print("Nadwaga")
else:
    print("Norma")
# warunki w if...else można zapisać w zmiennej (boolean)
```

```python
# 2. Przelicznik temperatury
temp_c = 23.5

temp_f = temp_c * 9 / 5 + 32
temp_k = temp_c + 273.15

print(f"""
Celsius: {temp_c:.2f}
Fahrenheit: {temp_f:.2f}
Kelvin: {temp_k:.2f}
""")
```

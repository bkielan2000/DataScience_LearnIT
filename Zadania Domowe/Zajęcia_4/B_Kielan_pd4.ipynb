```python
# Zadanie 12
samples = 50_000
batch_size = 128
epoch = 50
batch_time = 0.15

batches = samples // batch_size

if samples % batch_size != 0:
    batches += 1

total_batches = batches * epoch
total_time = total_batches * batch_time

hours = int(total_time // 3600)
minutes = int((total_time % 3600) // 60)
seconds = total_time % 60

print(f"Szacowany czas: {hours}h {minutes}m {seconds:.1f}s")
```

```python
# Zadanie 16
t_positives = 45
t_negatives = 38
f_positives = 7
f_negatives = 10

total_true_res = t_positives + t_negatives
total_false_res = f_positives + f_negatives
total_res = total_true_res + total_false_res

accuracy = total_true_res / total_res
precision = t_positives / (t_positives + f_positives)
recall = t_positives / (t_positives + f_negatives)
f1 = 2 * precision * recall / (precision + recall)

matrix = f"""
            Predicted
              0    1
Actual  0    {t_negatives}   {f_positives}
Actual  1    {t_positives}   {f_negatives}"""

print(matrix)
print(f"""
Metryki:
Accuracy: {accuracy:.3f}%
Precision: {precision:.3f}%
Recall: {recall:.3f}%
F1 Score: {f1:.3f}%
""")
```

```python
# Zadanie 20
table_name = "patients"
columns = ["patient_id", "age", "diagnosis"]
age_condition = 40
diagnoses_condition = ['diabetes', 'hypertension']
sort = "desc"
limit = 100

query = f"""
SELECT {", ".join(columns)}
    FROM {table_name}
    WHERE age > {age_condition}
    AND diagnosis IN ({diagnoses_condition})
    ORDER BY age {sort.upper()}
    LIMIT {limit}
"""

print(query)
```

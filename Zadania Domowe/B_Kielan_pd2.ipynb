# Wnioski z działaniach na listach

- Duże listy zajmują również coraz więcej pamięci operacyjnej, ponieważ wszystkie elementy muszą zostać zapisane w pamięci komputera
- Wraz ze wzrostem liczby elementów zużycie pamięci rośnie liniowo podobnie jak czas wykonania programu
- Wraz ze wzrostem liczby elementów listy zwiększa się czas potrzebny na obliczenie średniej
- Dla małych list czas wykonania jest znacznie niższy niż dla list z większą ilością danych
- Obliczanie średniej przy użyciu `sum(lista) / len(lista)` wymaga przejścia przez wszystkie elementy listy
- Wraz ze wzrostem liczby średnia wzrasta liniowo od `24.5` do `2499999.5`, tym większa lista tym więcej dodatkowych `9` zostaje dodanych między `24` a `.5`

# Pomysł na szybsze obliczanie średniej

Dla liczb: `0, 1, 2, ..., n-1` nie trzeba tworzyć całej listy gdyż tworzony jest ciąg arytmetyczny gdzie pierwsza liczba to `0` a ostatnia `n-1`
Z tego powodu średnią można zapisać jako `(pierwszy_element + ostatni_element) / 2`
Na przykład: dla `n = 10` mamy `(0 + 9) / 2` co daje poprawny wynik `4.5`

```python
num1 = 0
last_num = 49

start = time.time()

average = (num1 + last_num) / 2

end = time.time()
full_time = end - start

print(average)
print(full_time)
```

# Wnioski 'nowego' sposobu obliczania średniej
- Używając takiego sposobu nie wymagane jest tworzenie listy
- Z powodu nie tworzenia list zajmowane jest mniej pamięci
- jako że nie jest używana funkcja `sum()` i nie jest potrzebna iteracja po liście to nie zależnie od wielkości liczb obliczenie takiej średniej jest niskie i bardzo podobne między różnymi danymi
- tworzenie listy może mieć sens w sytuacji kiedy wykonywane są również inne działania na tych danych co prowadzi że nie jest potrzebne ponowne definiowanie zmiennych lub lista nie tworzy ciągu arytmetycznego
- własność matematyczna problemu praktycznie nie zużywa dodatkowej pamięci oraz nie zależnie od wielkości ciągu czas obliczeń jest tak na prawdę taka sama
